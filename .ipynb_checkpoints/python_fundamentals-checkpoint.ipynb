## Number Guessing Game 

## Level 1. print if the guessed number is right or not

```python
secret_number=56

user_input=int(input("Enter your guess :"))

if secret_number == user_input:
    print("Congratulations! you're guess is right")
else:
    print("Sorry, You're guess is wrong")
```

## Level 2. print if the guessed number is right or not with maximum of 7 trails

#### 2.1 using for loop

```python
secret_number=56

for i in range(0,7):
    user_input=int(input("Enter your guess :"))
    if secret_number == user_input:
        print("Congratulations! you're guess is right")
        break
    else:
        print("Sorry, You're guess is wrong")
    
```

#### 2.2 using while loop

```python
secret_number=56

while True:
    user_input=int(input("Enter your guess :"))
    if secret_number == user_input:
        print("Congratulations! you're guess is right")
        break
    else:
        print("Sorry, You're guess is wrong")
    
```

##  Level 3: Give multiple conditions

1.user should enter only number in range of 1 to 100 else "invalid number"

2.if the user guessed number is greater than the secret number the say "number is greater the secret number"

3.if the user guessed number is lesser than the secret number the say "number is lesser the secret number"

```python
secret_number=98
for i in range(1,8):
    user_input=int(input("Enter your guess"))
    
    if(user_input <0 or user_input>100):
       print("Invalid number!,please enter a number from 0 to 100")
    elif(user_input > secret_number):
       print("your number is greater than secret number")
    elif(user_input < secret_number):
       print("your number is lesser than secret number")
    else:
       print(f"Congradulations you guessed {secret_number} in {i} attempt")
       break
       
                    
                   
                   
               
```
