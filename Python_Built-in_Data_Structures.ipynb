## 1. List

- 1D
- hetrogeneous
- has duplicates
- indexing and slicing possible
- mutable

```python
# A 1D list -enclosed by square brackets , empty-list is allowed

fruits=["mango","apple","orange"]
print(fruits)
print(fruits[1])
print(type(fruits))
print("")

fruits1=[]
print(fruits1)
```

```python
# hetrogeneous-can have combination of different type of data

user_profile=["Sam",20,8.61,True,"placed"]
print(user_profile[0])
print(user_profile[1])
print(user_profile[2])
print(user_profile[3])
print(user_profile[4])
```

```python
# duplicates
fruits=["mango","apple","orange","mango","apple"]
print(fruits)
```

```python
# indexing -positive & negative

fruits=["mango","apple","orange","mango","grape","pomogranate","banana","litchy"]
print(fruits[-8])
print(fruits[-1])
print(fruits[-2:-1])
print(fruits[-2:1])
print(fruits[-2:-1])
print(fruits[0:7])
print(fruits[5:])
print(fruits[1:7:2])
print(fruits[2::])

```

```python
# mutalble- list can be changed 
fruits=["mango","apple","orange","mango","apple"]
fruits.append("grapes")

fruits[3]="watermelon";

fruits

```

## 2. Sets
- 1D
- hetrogenous
- doesn't allow duplicates
- doesn't support indexing and slicing
- mutable (only using functions not by indexing)

```python
# A 1D set -enclosed by curly braces , empty-list is allowed
user_profile={"Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True}
user_profile2={}

print(user_profile)
print(user_profile2)

```

```python
# no indexing & slicing
user_profile={"Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True}
print(user_profile[2])
print(user_profile[2:6])
```

```python
# mutable using functions
user_profile={"Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True}
print(user_profile.append("JOHN"))
      

```

```python
# mutable using functions add()-adds in ascending order
user_profile={"Sam",20,8.61,True,"chennai","03.01.2006",100,20,8.61,True}
user_profile.add("Ambattur")

user_profile
      

```

```python
# mutable using functions
user_profile={"Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True}
user_profile.discard(100)
user_profile
      

```

## 3.Tuples
- 1D
- hetrogenous
- allows duplicates
- Indexing & slicing posiible
- Immutable

```python
# A 1D tuple -enclosed by paranthesis , empty-list is allowed
user_profile=("Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True,"John")
user_profile2=();

print(user_profile)
print(user_profile2)

```

```python
#  if a collection of data without any brackets it is recogonized as tuple
user_profile="Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True,"John"
type(user_profile)
```

```python
#  if a collection of data without any brackets it is recogonized as tuple
user_profile="Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True,"John"

print(user_profile)
```

```python
#indexing & slicing are allowed

user_profile=("Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True,"John")
print(user_profile[0])
print(user_profile[-11:-5])


```

```python
# immutable using indexing and also functions (can't assign any variables NEW inside tuple)

user_profile=("Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True,"John")
print(user_profile.add("ball"))
print(user_profile[0]="Blessing)


```

```python
# some methods of tuple

user_profile=("Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True,"John")
user_profile.count(8.61)
```

```python
# some methods of tuple

user_profile=("Sam",20,8.61,True,"Ambattur","03.01.2006",100,20,8.61,True,"John")
user_profile.index(100)
```
